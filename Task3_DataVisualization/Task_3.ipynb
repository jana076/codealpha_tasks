Imports and setup

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_style("darkgrid")
```

Load and clean the dataset

```python
df = sns.load_dataset("titanic").dropna(subset=["age", "fare"])
```

Create the dashboard figure (2x2 grid)

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 10))
fig.suptitle(
    "Titanic Data Analysis - Mini Dashboard", fontsize=16, color="navy"
)

# Plot 1: Survival Rate by Class
sns.barplot(
    data=df, x="pclass", y="survived", ax=axes[0, 0], palette="viridis"
)
axes[0, 0].set_title("Survival Rate by Passenger Class")
axes[0, 0].set_xlabel("Passenger Class")
axes[0, 0].set_ylabel("Survival Rate")

# Plot 2: Age Distribution
sns.histplot(
    data=df,
    x="age",
    hue="survived",
    multiple="stack",
    ax=axes[0, 1],
    palette="coolwarm",
)
axes[0, 1].set_title("Age Distribution by Survival Status")

# Plot 3: Age vs Fare
sns.scatterplot(
    data=df,
    x="age",
    y="fare",
    hue="survived",
    ax=axes[1, 0],
    palette="cool",
    alpha=0.7,
)
axes[1, 0].set_title("Age vs Fare")

# Plot 4: Fare Distribution by Class
sns.boxplot(data=df, x="pclass", y="fare", ax=axes[1, 1], palette="Set2")
axes[1, 1].set_title("Fare Distribution by Passenger Class")

# Layout Adjustments & Export
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.savefig("dashboard.png", dpi=150)
plt.show()
```
